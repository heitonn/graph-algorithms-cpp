# Plotting runtimes


```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("benchmark_results.csv")
```

```python
# plotting runtime for tarjans scc

scc = df[df["algorithm"] == "scc"]

for rep in ["incidence", "matrix"]:
    sub = scc[scc["representation"] == rep]
    plt.plot(sub["n"], sub["avg_time_us"], marker="o", label=rep)

plt.xlabel("Number of nodes n")
plt.ylabel("Average runtime (microseconds)")
plt.title("Tarjan SCC runtime")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Diamond, constant m

diamond_const = df[(df["algorithm"] == "diamond") & (df["case"] == "constant_m")]

for rep in ["incidence", "matrix"]:
    sub = diamond_const[diamond_const["representation"] == rep]
    plt.plot(sub["n"], sub["avg_time_us"], marker="o", label=rep)

plt.xlabel("Number of nodes n")
plt.ylabel("Average runtime (microseconds)")
plt.title("Diamond query runtime (constant m = 3)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Diamond, m proportional to n

diamond_prop = df[(df["algorithm"] == "diamond") & (df["case"] == "proportional_m")]

for rep in ["incidence", "matrix"]:
    sub = diamond_prop[diamond_prop["representation"] == rep]
    plt.plot(sub["n"], sub["avg_time_us"], marker="o", label=rep)

plt.xlabel("Number of nodes n")
plt.ylabel("Average runtime (microseconds)")
plt.title("Diamond query runtime (m proportional to n)")
plt.legend()
plt.grid(True)
plt.show()
```
